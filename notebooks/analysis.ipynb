```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/superstore.csv", encoding="latin-1")
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)
df.head()
```

```python
df.columns
```

```python
top_sub = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)
top_sub.plot(kind="barh", figsize=(8, 5))
plt.title("Top 10 Sub-Categories by Sales")
plt.xlabel("Total Sales")
plt.gca().invert_yaxis()
plt.show()
```

**Phones and Chairs leands the sales by Sub-Categories**

```python
df.groupby("Region")["Sales"].sum().plot(kind="pie", autopct="%1.0f%%", figsize=(6, 6))
plt.title("Share of Sales by Region")
plt.ylabel("")
plt.show()
```

**West and East holds the most share of sales**

```python
sns.barplot(data=df, x="Segment", y="Sales", estimator="sum", errorbar=None)
plt.title("Total Sales by Customer Segment")
plt.show()
```

**Consumers are the total sales by Customer Segment**

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Ship Mode", y="Days to Ship",
            order=["Same Day", "First Class", "Second Class", "Standard Class"],
            palette="Blues")
plt.title("Shipping Time by Ship Mode")
plt.show()
```

**Same Day Delivery is the fastest way of shipping time, It leads by a huge margin**

```python
df["Year"] = df["Order Date"].dt.year
yearly = df.groupby(["Year", "Category"])["Sales"].sum().unstack()
yearly.plot(kind="line", marker="o", figsize=(8, 5))
plt.title("Yearly Sales by Category")
plt.ylabel("Total Sales")
plt.show()
```

**Technology grew fastest based on yearly sales by category**
